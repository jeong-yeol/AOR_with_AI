# Panel Regression

Fixed-effects panel regression on the business-month panel.

```python
import pandas as pd
from linearmodels.panel import PanelOLS
panel = pd.read_csv('sample_business_month_panel.csv')
panel = panel.set_index(['business_id','year_month'])
m = PanelOLS(panel['avg_rating'], panel[['avg_sentiment','review_count']], entity_effects=True).fit(cov_type='robust')
print(m)
```
